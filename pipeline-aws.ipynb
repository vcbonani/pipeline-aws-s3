```python
#importar biblioteca requests para permitir acessar os dados do Data Boston
import urllib.request
import requests
import pandas as pd
```

```python
#criar função que vai obter os arquivos com os dados
def extrair_dados(url, nome_arquivo):
  #define o header para requisitar a api
  #o user-agent informa qual é o browser requisitante, e o accept informa o tipo do arquivo
  headers = {'User-Agent':'Mozilla/5.0', 'Accept':'text/csv'}
  #passa os parâmetros
  response = requests.get(url, headers = headers, stream = True)
  #verifica se o status code é 200, ou seja se obteve sucesso na requisição
  if response.status_code == 200:
    #abre um arquivo para escrita em binário
    with open (nome_arquivo, 'wb') as arquivo:
      #quebra o response em pequenos pedaços para gravar no arquivo => necessário fazer isso
      for chunk in response.iter_content(chunk_size=8192):
        #grava os pequenos pedaços de dados no arquivo
        arquivo.write(chunk)
  else:
    print('Erro: ', response.status_code)
```

```python
arquivos = [ 'data/dados_2015.csv', 'data/dados_2016.csv', 'data/dados_2017.csv', 'data/dados_2018.csv'
    , 'data/dados_2019.csv', 'data/dados_2020.csv' ]
```

```python
#cria um dataframe
dfs = {}
```

```python
#itera a lista com os nomes dos arquivos
for arquivo in arquivos:
  #obtém o ano a partir do nome do arquivo, entre o _ e o .
  ano = arquivo.split('_')[-1].split('.')[0]
  #armazena no data frame cada arquivo
  dfs[ano] = pd.read_csv(arquivo)
```

```python
#importar biblioteca boto3
import boto3

chave_acesso_aws = 'sua_chave'
chave_acesso_secreta_aws = 'sua_chave_secreta'
regiao_nome = 'us-east-1'

#configurar sessão padrão do boto3
boto3.setup_default_session(
    aws_access_key_id = chave_acesso_aws,
    aws_secret_access_key = chave_acesso_secreta_aws,
    region_name = regiao_nome
)

#criando cliente s3
s3 = boto3.client('s3')
```

```python
#criar arquivo para verificar a conexão
conteudo = 'Olá, S3'

#gera arquivo com o conteúdo gravado anteriormente
with open('oi-s3.txt', 'w+') as arquivo:
  arquivo.write(conteudo)
```

### Salvar arquivo em Parquet para melhorar a performance do arquivo

```python
#importando BytesIO
from io import BytesIO

#percorre a df gerada
for ano, df in dfs.items():
  #cria um objeto de BytesIO
  parquet_buffer = BytesIO()
  #converte a df para o tipo parquet
  df.to_parquet(parquet_buffer)

  #grava no bucket o objeto convertido para parquet
  s3.put_object(
      Bucket = 'alura-data-lake-aws-1', #informa o bucket
      Key = f'bronze/dados_{ano}.parquet', #informa o caminho e nome do arquivo
      Body = parquet_buffer.getvalue() #parâmetro do body
  )
```

```python
#listando os objetos
response = s3.list_objects(Bucket = 'alura-data-lake-aws-1')

#percorrendo o conteúdo do response para obter o nome dos objetos
keys = [obj['Key'] for obj in response ['Contents']]
keys
```
